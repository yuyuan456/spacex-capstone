# Data Wrangling - SpaceX Falcon 9 Launches

Objectives:
1. Inspect missing values and clean the dataset
2. Convert `LandingOutcome` into the binary target `Class`
3. Export the modelling dataset

```python
import pandas as pd

df = pd.read_csv("data/spacex_launch_data.csv")
print(df.shape)
df.head()
```

```python
df.isnull().sum()
```

```python
print(df["LandingOutcome"].value_counts())
```

```python
# 1 = successful first-stage landing, 0 = failure / no attempt
df["Class"] = df["LandingOutcome"].apply(lambda x: 1 if str(x).startswith("Success") else 0)
print(df["Class"].value_counts())
print()
print(df[["LandingOutcome", "Class"]].drop_duplicates().to_string(index=False))
```

```python
df.to_csv("dataset_part_3.csv", index=False)
print("Saved dataset_part_3.csv -", df.shape)
```
