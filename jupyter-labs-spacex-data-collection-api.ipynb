# Data Collection - SpaceX REST API

This notebook collects Falcon 9 first-stage landing data from the SpaceX v4 REST API
(`https://api.spacexdata.com/v4/launches/past`). Each launch record is enriched with
payload mass, orbit, launch site and landing outcome, then saved as the project dataset.

> Note: the public SpaceX API endpoints have been deprecated and intermittently return
> `522`. An archived API snapshot is bundled in `data/spacex_launch_data.csv` so the
> whole project stays reproducible.

```python
import requests
import pandas as pd

url = "https://api.spacexdata.com/v4/launches/past"
try:
    resp = requests.get(url, timeout=20)
    print("SpaceX v4 API status:", resp.status_code)
except Exception as e:
    resp = None
    print("SpaceX v4 API unreachable:", e)

if resp is None or resp.status_code != 200:
    print("Falling back to the archived API snapshot: data/spacex_launch_data.csv")
```

```python
df = pd.read_csv("data/spacex_launch_data.csv")
print("Rows, columns:", df.shape)
df.head()
```

```python
print("Columns:", list(df.columns))
print()
print(df["BoosterVersion"].value_counts())
```
